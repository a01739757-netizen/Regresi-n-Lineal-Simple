```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
archivo = "Analisis GAC Full7agosto 2026(Leads Reales).csv"

df = pd.read_csv(archivo, skiprows=2, nrows=29, thousands=",", encoding="utf-8-sig")
df
```

```python
# Borrar las 2 columnas vacías
df = df.drop(columns=["Unnamed: 12", "Unnamed: 13"])

# Quitar espacios de los nombres de las columnas
df.columns = df.columns.str.strip()

df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Año"] = df["Año"].ffill()
df["Mes"] = df["Mes"].str.strip()

df.head(15)
```

```python
for columna in ["Conversion", "Efectividad de Leads"]:
    df[columna] = df[columna].str.replace("%", "").astype(float)

df.head()
```

```python
# Filas que tienen al menos una celda vacía
df[df.isnull().any(axis=1)]
```

```python
print("Filas antes de quitar nulos:", df.shape[0])

df = df.dropna()                 # elimina las filas con celdas vacías
df = df.reset_index(drop=True)   # reordena los índices (0, 1, 2, ...)

# Ahora que ya no hay celdas vacías, pasamos los conteos a números enteros
for columna in ["Total", "Ilocalizable", "No Contesta", "D. Incorrectos",
                "Duplicados", "Efectivos", "Abiertos", "Ventas"]:
    df[columna] = df[columna].astype(int)

print("Filas después de quitar nulos:", df.shape[0])
```

```python
columnas_conteo = ["Total", "Ilocalizable", "No Contesta", "D. Incorrectos",
                   "Duplicados", "Efectivos", "Abiertos", "Ventas"]

df[columnas_conteo].boxplot(figsize=(12, 5), rot=45)
plt.title("Diagrama de caja ANTES de quitar atípicos")
plt.show()
```

```python
# Empezamos suponiendo que TODAS las filas se quedan (True)
filas_ok = pd.Series(True, index=df.index)

for columna in columnas_conteo:
    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    # ¿Qué filas están fuera de los límites en esta columna?
    es_atipico = (df[columna] < limite_inferior) | (df[columna] > limite_superior)

    print(f"{columna}: límites [{limite_inferior:.0f} , {limite_superior:.0f}]  -> atípicos: {es_atipico.sum()}")

    # Si una fila es atípica en esta columna, ya no se queda
    filas_ok = filas_ok & (~es_atipico)
```

```python
# Ver los meses que se van a eliminar por atípicos
df[~filas_ok][["Año", "Mes"] + columnas_conteo]
```

```python
print("Filas antes de quitar atípicos:", df.shape[0])

df = df[filas_ok]                # nos quedamos solo con las filas buenas
df = df.reset_index(drop=True)

print("Filas después de quitar atípicos:", df.shape[0])
```

```python
df[columnas_conteo].boxplot(figsize=(12, 5), rot=45)
plt.title("Diagrama de caja DESPUÉS de quitar atípicos")
plt.show()
```

```python
# La columna "Año" ya se imputó en el Paso 5 con ffill (el año que corresponde a cada mes).
# Aquí solo la dejamos como número entero.
df["Año"] = pd.to_numeric(df["Año"]).astype(int)

df["Año"].head()
```

```python
# Diccionario: nombre del mes -> número del mes
meses = {
    "Enero": 1, "Febrero": 2, "Marzo": 3, "Abril": 4,
    "Mayo": 5, "Junio": 6, "Julio": 7, "Agosto": 8,
    "Septiembre": 9, "Octubre": 10, "Noviembre": 11, "Diciembre": 12
}

df["Mes"] = df["Mes"].map(meses)      # reemplaza el texto por el número
df["Mes"] = pd.to_numeric(df["Mes"])  # se asegura de que sea numérico

df.head()
```

```python
print(df.dtypes)

df.to_csv("Leads Reales Limpio.csv", index=False, encoding="utf-8-sig")
print("\nArchivo guardado: Leads Reales Limpio.csv")
```

```python
# Los 8 tipos de Leads
columnas_leads = ["Total", "Ilocalizable", "No Contesta", "D. Incorrectos",
                  "Duplicados", "Efectivos", "Abiertos", "Ventas"]

# Matriz de correlación (cada columna contra cada columna)
correlaciones = df[columnas_leads].corr()
correlaciones
```

```python
# Mapa de calor (heat map) de las correlaciones
plt.figure(figsize=(9, 7))
sns.heatmap(correlaciones, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Mapa de calor de correlación entre tipos de Leads")
plt.show()
```

```python
# El valor absoluto quita el signo: sirve para ver solo QUÉ TAN FUERTE es la relación
correlaciones_abs = correlaciones.abs()

plt.figure(figsize=(9, 7))
sns.heatmap(correlaciones_abs, annot=True, fmt=".2f", cmap="Greens")
plt.title("Fuerza de la correlación (valor absoluto)")
plt.show()
```

```python
# Un diagrama de dispersión de cada tipo de Lead contra el Total
otros_tipos = ["Ilocalizable", "No Contesta", "D. Incorrectos",
               "Duplicados", "Efectivos", "Abiertos", "Ventas"]

plt.figure(figsize=(14, 7))
for i, columna in enumerate(otros_tipos):
    plt.subplot(2, 4, i + 1)                        # cuadrícula de 2 filas x 4 columnas
    sns.scatterplot(x=columna, y="Total", data=df)
    plt.title(columna + " vs Total")

plt.tight_layout()
plt.show()
```
